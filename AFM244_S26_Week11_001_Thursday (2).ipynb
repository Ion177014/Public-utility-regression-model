```python
import matplotlib.pyplot as plt
import statsmodels.api as sm
import numpy as np
import pandas as pd

pd.options.display.float_format = '{:.2f}'.format

```

```python
qSales = pd.read_csv('qSales_2024.csv')

display(qSales)
```

```python
#to do time series forecasting
# Step 0 is always to visualize the data to identify any patterns
# step 1 is to introduce time as one of our components to the model
# step 2 introduce any dummy variables you might find
# step 3 is to do multiple regression modelling

#our focus is to prepare a time series model for apple

```

```python
#change date to datetime type

qSales['datadate'] = pd.to_datetime(qSales['datadate'])
```

```python
apple_sales = qSales.loc[qSales['tic']=='AAPL']

display(apple_sales)
```

```python
plt.figure(figsize=(10,6))

plt.title('apple revenue')

plt.xlabel('date')
plt.ylabel('revenue')

plt.plot(apple_sales['datadate'],apple_sales['saleq'],marker='o')
```

```python
#to introduce the time component of time series regession forecasting
# all we have to do is add a new column - call it time - and measure incrementally increasing time periods (1, 2, 3, ...)

#base requirement is that each row represents an equal block of time
 # (each row is a day or a quarter, etc.)

apple_sales['time'] = range(1, len(apple_sales) + 1)
#be sure to add the + 1

display(apple_sales)
```

```python
# this is how to take 75% and 25% of the dataset

dt4training = apple_sales[:int(0.75 * len(apple_sales))] #takes the first 75% of rows
dt4testing = apple_sales[int(0.75 * len(apple_sales)):] #takes the last 25% of rows

```

```python
y = dt4training.loc[:,'saleq']
x = dt4training.loc[:,'time']

x = sm.add_constant(x)

model1 = sm.OLS(y,x).fit()

model1.params
```

#apple revenue = -13,536 + 1077.61 * time

```python
# how to forecast a time series model
# time series is much less precise than regression normally
```

#during time series forecasting its best to provide a range
# luckily python can take care of that for us

```python
# to get a range of values we do the following
#create new dataframe with prediction results
# get summary frame with a confidence level

model1_prediction_results = model1.get_prediction(x)
prediction_intervals = model1_prediction_results.summary_frame(alpha=0.2) # 80% confidence interval

display(prediction_intervals)
```

```python
# option 1 for our dummy variable is to check for fqtr == 1
# option 2 for our dummy variable is to list out the dates into a list
# release_dates = ['2011-12-31', '2012-12-31']

apple_sales['iphone_dv'] = np.where(apple_sales['fqtr']==1,1,0)
apple_sales['iphone_interaction'] = apple_sales['time'] * apple_sales['iphone_dv']

display(apple_sales)
```

```python
# this is how to take 75% and 25% of the dataset

dt4training = apple_sales[:int(0.75 * len(apple_sales))] #takes the first 75% of rows
dt4testing = apple_sales[int(0.75 * len(apple_sales)):] #takes the last 25% of rows
```

```python
x = dt4training.loc[:,['time','iphone_dv','iphone_interaction']]
x = sm.add_constant(x)

y = dt4training.loc[:,'saleq']

model_dummy = sm.OLS(y,x).fit()

model_dummy.params
```

apple's revenue = -11,044 + 933 * time + -10422 * iphone_dv + 578*(time * iphone_dv)

```python
# to forecast values
#first prepare the testing independent variables (time, dummy, interaction)
# then use the get_prediction function to forecast the values into the testing set

x_test = dt4testing.loc[:,['time','iphone_dv','iphone_interaction']]
x_test = sm.add_constant(x_test)
```

```python
model_dummy.get_prediction(x_test).summary_frame(alpha=0.2)
model_dummy.get_prediction(x_test).summary_frame(alpha=0.1)
```
